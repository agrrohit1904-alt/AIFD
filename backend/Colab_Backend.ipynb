```python
# Download Cloudflare Tunnel
!wget -q -c -nc https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64
!chmod +x cloudflared-linux-amd64

import subprocess
import time

# Start the FastAPI server in the background
print("Starting FastAPI server...")
subprocess.Popen(["uvicorn", "app.main:app", "--host", "0.0.0.0", "--port", "8000"])
time.sleep(3) # Wait for server to start

# Start Cloudflare Tunnel
print("Starting Cloudflare Tunnel...")
p = subprocess.Popen(["./cloudflared-linux-amd64", "tunnel", "--url", "http://127.0.0.1:8000"], stderr=subprocess.PIPE, text=True)

# Extract and print the URL
for line in p.stderr:
    if "trycloudflare.com" in line:
        url = line[line.find("https"):]
        print("\n" + "="*60)
        print("\u2705 YOUR SECURE PUBLIC API URL:")
        print(url.strip())
        print("="*60 + "\n")
        break

# Keep the notebook running
import IPython
display(IPython.display.HTML("<h2 style='color:green;'>Server is running! Copy the URL above!</h2>"))
while True:
    time.sleep(60)
```
